### Municipalities
Focus on two main Municipalities in `São Paulo`, `Rio de Janeiro`.
They have the largest order volumne, and two of them account for a little more than 20% of total volume.

```python
import pandas as pd
import geopandas as gpd
import folium
import matplotlib.pyplot as plt
import seaborn as sns

from utils import read_customer_seller
```

```python
def get_state(df: pd.DataFrame, state: str) -> pd.DataFrame:
    final_cols = []
    
    state_df = df[df["cus_state"] == state]
    
    state_df.drop_duplicates(subset=['order_id'])
    
    res = state_df[['order_id', 'cus_lat', 'cus_lng']].copy()
    res['delivery_hours'] = (state_df['order_delivered_customer_date'] - state_df['order_purchase_timestamp']).dt.total_seconds() / 3600
    res.columns = ['order_id', 'lat', 'lng', 'delivery_hours']
    return res
```

```python
def read_gdf(state: str) -> gpd.GeoDataFrame:
    return gpd.read_file(f"../data/others/geojson/data/{state}.json")
```

```python
def convert_gdf(df: pd.DataFrame) -> gpd.GeoDataFrame:
    geometry = gpd.points_from_xy(df.lng, df.lat)
    gdf = gpd.GeoDataFrame(
        df,
        geometry=geometry,
        crs="EPSG:4326"
    )
    return gdf
```

```python
def join_n_filter(df: pd.DataFrame, geo: gpd.GeoDataFrame, geocode : str) -> gpd.GeoDataFrame:
    df = gpd.sjoin(df, geo, predicate = "within", how='right')
    
    df.drop_duplicates(subset=['order_id'])   
    df = df[df['GEOCODIGO'] == geocode]
    return df
```

```python
# Rank orders by delivery_hours
# 0 is Fastest, and 4 is the Slowest
def rank_orders(df: pd.DataFrame) -> pd.DataFrame:
    df = df.dropna().copy()
    
    df['rank'] = pd.qcut(df['delivery_hours'], 
                                   q=5,
                                   labels= False)
    return df
```

```python
def transform(df: pd.DataFrame, state: str,  geocode : str) -> gpd.GeoDataFrame:
    df = get_state(df, state)
   
    df = convert_gdf(df)
    df = join_n_filter(df, read_gdf(state), geocode)
    df = rank_orders(df)
    
    return df
```

```python
df = read_customer_seller()
df.head()
```

```python
sp = transform(df, 'SP', '3550308')
sp.head()
```

```python
# Center point for Nashville
sao_paulo = [-23.5558, -46.6396]

# Create map
m = folium.Map(location=sao_paulo, zoom_start=12, tiles='CartoDB positron')

# 3. Define Colors for Categories
colors = ['green', 'green', 'green', 'green', 'red']

# 4. Add Points to Map
for i, row in sp.iterrows():
    folium.CircleMarker(
        location=[row['lat'], row['lng']],
        radius=0.5,
        color=colors[row['rank']],
        fill=True,
        fill_color=colors[row['rank']],
        fill_opacity=0.7
    ).add_to(m)

# Add Marker to Mark Center Hub
marker = folium.Marker(location = sao_paulo)
marker.add_to(m)
   
# Display the map
folium.LayerControl().add_to(m)
display(m)
```
